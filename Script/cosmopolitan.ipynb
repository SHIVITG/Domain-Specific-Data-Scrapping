```python
driver = webdriver.Chrome('chromedriver_linux64/chromedriver')
driver.get("https://www.cosmopolitan.com/style-beauty/fashion/")
```

```python
def get_link():
    soup = BeautifulSoup(driver.page_source)
    for i in soup.findAll('div',{'class': 'feed feed-list'}):
        for items in i('a'):
            link1 = "https://www.cosmopolitan.com" + items.get('href')
            print(link1)
            link.append(link1)
```
