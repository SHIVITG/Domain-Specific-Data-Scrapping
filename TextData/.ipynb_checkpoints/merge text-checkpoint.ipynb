```python
import os
files = []
for i in os.listdir("/home/shivi/Downloads/notebook/Data Files/"):
    if i.endswith('.txt'):
        files.append(i)
        num_words = 0
        with open(i, 'r') as f:
            for line in f:
                words = line.split()
                num_words += len(words)
                print("Number of words:")
                print(num_words)

with open('All_merge.txt', 'w') as outfile:
    for fname in files:
        with open(fname) as infile:
            for line in infile:  
                outfile.write(line)
```
