```python
driver = webdriver.Chrome('chromedriver_linux64/chromedriver')
driver.get("https://www.teenvogue.com/fashion")
```

```python
def get_link():
    soup = BeautifulSoup(driver.page_source)
    for i in soup.findAll('div',{'class':'row'}):
        for j in i.findAll('a',{'class':'feature-item-link feature-item-link-hed'}):
            link1 = "http:///www.teenvogue.com" + j.get('href')
            print(link1)
            link.append(link1)
   
```
