```python
# to fetch article link
def get_link():
    soup = BeautifulSoup(driver.page_source)
    for i in soup.findAll('section',{'class': 'article-list'}):
        for items in i('a'):
            link1 ="https://www.telegraph.co.uk" + items.get('href')
            #print(link1)
            link.append(link1)
```
